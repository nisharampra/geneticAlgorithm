```python
import random

# --------- Problem: Maximize f(x) = x^2 ---------
def fitness(x):
    return x ** 2

# Convert binary to integer
def decode(chromosome):
    return int(chromosome, 2)

# Generate random population
def generate_population(size, chromosome_length):
    return [''.join(random.choice('01') for _ in range(chromosome_length)) for _ in range(size)]

# Select best N individuals (elitism selection)
def select(population, num_parents):
    sorted_population = sorted(population, key=lambda x: fitness(decode(x)), reverse=True)
    return sorted_population[:num_parents]

# Crossover (single-point)
def crossover(parents, offspring_size):
    offspring = []
    for _ in range(offspring_size):
        p1, p2 = random.sample(parents, 2)
        point = random.randint(1, len(p1)-1)
        child = p1[:point] + p2[point:]
        offspring.append(child)
    return offspring

# Mutation (bit flip)
def mutate(chromosome, mutation_rate):
    return ''.join(
        bit if random.random() > mutation_rate else '1' if bit == '0' else '0'
        for bit in chromosome
    )

# Genetic Algorithm
def genetic_algorithm(pop_size=10, chromosome_length=6, generations=20, mutation_rate=0.01):
    population = generate_population(pop_size, chromosome_length)
    
    for gen in range(generations):
        decoded = [decode(c) for c in population]
        best = max(population, key=lambda x: fitness(decode(x)))
        print(f"Gen {gen+1} | Best: {decode(best)} | f(x)={fitness(decode(best))}")
        
        parents = select(population, num_parents=4)
        offspring = crossover(parents, pop_size - len(parents))
        offspring = [mutate(child, mutation_rate) for child in offspring]
        
        population = parents + offspring

    # Final best
    best_solution = max(population, key=lambda x: fitness(decode(x)))
    return decode(best_solution), fitness(decode(best_solution))

# Run
best_x, best_fx = genetic_algorithm()
print(f"\nBest solution: x={best_x}, f(x)={best_fx}")
```
